```python
import pandas as pd
import numpy as np



# ============================================================
# 1) HANDLE DATE COLUMN: album_release_date → datetime
# ============================================================
if 'album_release_date' in df.columns:
    df['album_release_date'] = pd.to_datetime(df['album_release_date'], errors='coerce')
    # Fill missing with earliest valid date
    df['album_release_date'] = df['album_release_date'].fillna(df['album_release_date'].min())


# ============================================================
# 2) NUMERIC COLUMNS: convert → fill median
# ============================================================
numeric_cols = [
    'track_popularity', 'artist_popularity',
    'artist_followers', 'track_duration_min',
    'album_total_tracks'
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(df[col].median())


# ============================================================
# 3) CATEGORICAL COLUMNS: fill missing with mode
# ============================================================
categorical_cols = [
    'track_name', 'artist_name',
    'album_name', 'artist_genres',
    'album_type'
]

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])


# ============================================================
# 4) PROCESS GENRE COLUMN: multi-genre → list
# ============================================================
if 'artist_genres' in df.columns:
    df['artist_genres'] = df['artist_genres'].replace('', 'Unknown')
    df['artist_genres_list'] = df['artist_genres'].astype(str).str.split(r'[;,]\s*', regex=True)
else:
    df['artist_genres_list'] = [['Unknown']] * len(df)


# ============================================================
# 5) CREATE YEAR COLUMN
# ============================================================
if 'album_release_date' in df.columns:
    df['Year'] = df['album_release_date'].dt.year
else:
    df['Year'] = np.nan


# ============================================================
# 6) FILL TRACK ID IF MISSING (rare)
# ============================================================
if 'track_id' in df.columns:
    df['track_id'] = df['track_id'].fillna('unknown_track')


# ============================================================
# SUMMARY
# ============================================================
print("Preprocessing complete.")
print("Shape:", df.shape)
print("Columns:", df.columns.to_list())
```

🟦 Q1. How have global music streaming trends evolved over the years (2009–2025)?

```python

# Q1 : Average Track Popularity Over Time (2009–2025)

# 1) Create yearly summary statistics
ts_df = (
    df.dropna(subset=['Year'])
      .groupby('Year', as_index=False)
      .agg(
          avg_track_popularity=('track_popularity', 'mean'),
          median_track_popularity=('track_popularity', 'median'),
          count_tracks=('track_id', 'count')
      )
)

# 2) Restrict to 2009–2025
ts_df = ts_df[(ts_df['Year'] >= 2009) & (ts_df['Year'] <= 2025)]
ts_df = ts_df.sort_values('Year')

# 3) Main line chart
fig_q1 = px.line(
    ts_df,
    x='Year',
    y='avg_track_popularity',
    markers=True,
    title="Q1 — Average Track Popularity by Year (2009–2025)",
    labels={'avg_track_popularity': 'Avg Track Popularity'}
)

# 4) Add track count bar trace
fig_q1.add_bar(
    x=ts_df['Year'],
    y=ts_df['count_tracks'],
    name='Track Count',
    opacity=0.25,
    marker_line_width=0
)

# 5) Update layout
fig_q1.update_layout(
    template='plotly_white',
    height=500,
    yaxis_title="Avg Track Popularity",
    xaxis_title="Year"
)

# 6) Update ONLY the line traces (avoid messing up the bar)
fig_q1.update_traces(
    mode='lines+markers',
    selector=dict(type='scatter')
)

# 7) Show chart
fig_q1.show()
```

🟨 Q2. Which genres dominate the global music landscape, and how are genres distributed across tracks?

```python
# Q2: Genre -> Artist -> Track Sunburst
import pandas as pd
import plotly.express as px

# Load (if you want to re-load; skip if df already in memory)
# df = pd.read_csv("/mnt/data/spotify_data clean.csv")

# Ensure genre list column exists (created during preprocessing as artist_genres_list)
if 'artist_genres_list' not in df.columns and 'artist_genres' in df.columns:
    df['artist_genres_list'] = df['artist_genres'].fillna('Unknown').astype(str).str.split(r'[;,]\s*', regex=True)

# Explode into one row per (genre, artist, track)
df_sb = df.explode('artist_genres_list').copy()
df_sb['artist_genres_list'] = df_sb['artist_genres_list'].str.strip().replace('', 'Unknown').fillna('Unknown')

# Create a small count per track (1) used as value
df_sb['count'] = 1

# optional: keep top K genres for readability
top_genres = df_sb['artist_genres_list'].value_counts().head(15).index.tolist()
df_sb_small = df_sb[df_sb['artist_genres_list'].isin(top_genres)].copy()

fig_q3 = px.sunburst(
    df_sb_small,
    path=['artist_genres_list', 'artist_name', 'track_name'],
    values='count',
    maxdepth=3,
    title="Q3 — Genre → Artist → Track (Top genres)",
    template='plotly_white'
)
fig_q3.update_traces(textinfo='label+percent entry')
fig_q3.update_layout(height=700)
fig_q3.show()
```

🟧 Q3. How does track popularity change over time for the top artists or top tracks?

```python
# Q4: Multi-line time-series of avg track popularity for top artists with dropdown
import plotly.graph_objects as go
import numpy as np

# Make sure Year present
if 'Year' not in df.columns and 'album_release_date' in df.columns:
    df['album_release_date'] = pd.to_datetime(df['album_release_date'], errors='coerce')
    df['Year'] = df['album_release_date'].dt.year

artist_year = (
    df.dropna(subset=['Year'])
      .groupby(['artist_name','Year'], as_index=False)
      .agg(avg_track_popularity=('track_popularity','mean'), releases=('track_id','count'))
)

# Choose candidate artists (top by max followers)
artist_rank = df.groupby('artist_name', as_index=False).agg(max_followers=('artist_followers','max'))
artist_rank = artist_rank.sort_values('max_followers', ascending=False).head(20)
artist_candidates = artist_rank['artist_name'].tolist()

# Build figure: one trace per candidate artist (initially show top N)
N_init = 6
visible_init = [i < N_init for i in range(len(artist_candidates))]

fig_q4 = go.Figure()
for i, artist in enumerate(artist_candidates):
    d = artist_year[artist_year['artist_name'] == artist].sort_values('Year')
    fig_q4.add_trace(go.Scatter(
        x=d['Year'],
        y=d['avg_track_popularity'],
        mode='lines+markers',
        name=artist,
        visible=visible_init[i],
        hovertemplate='%{x}<br>%{y:.2f}<br>' + artist
    ))

# Dropdown buttons
buttons = []
options = [3, 6, 10, len(artist_candidates)]
for opt in options:
    vis = [i < opt for i in range(len(artist_candidates))]
    buttons.append(dict(label=f"Top {opt}", method="update", args=[{"visible": vis},
               {"title": f"Q4 — Avg Track Popularity Over Time (Top {opt} artists)"}]))

buttons.append(dict(label="Show all", method="update", args=[{"visible": [True]*len(artist_candidates)},
               {"title": "Q4 — Avg Track Popularity Over Time (All candidates)"}]))

fig_q4.update_layout(
    updatemenus=[dict(active=1, buttons=buttons, x=0.0, y=1.12)],
    title=f"Q4 — Avg Track Popularity Over Time (Top {N_init} artists shown)",
    xaxis_title="Year",
    yaxis_title="Avg Track Popularity",
    template='plotly_white',
    height=600
)
fig_q4.show()
```

Q5 🟦 Q5. How strongly are the numeric features correlated with each other, and what key relationships can be identified from a correlation heatmap?

```python
# Q5: Correlation heatmap for numeric features
import plotly.graph_objects as go
corr_cols = [c for c in ['track_popularity','artist_popularity','artist_followers','track_duration_min','album_total_tracks'] if c in df.columns]

if len(corr_cols) < 2:
    print("Not enough numeric columns for correlation heatmap.")
else:
    corr = df[corr_cols].astype(float).corr().round(3)
    z = corr.values

    heatmap_q6 = go.Figure(data=go.Heatmap(
        z=z,
        x=corr.columns,
        y=corr.index,
        text=corr.values,
        texttemplate="%{text}",
        hoverongaps=False,
        colorscale='Viridis'
    ))

    heatmap_q6.update_layout(
        title="Q6 — Correlation Heatmap (numeric features)",
        template='plotly_white',
        height=500
    )
    heatmap_q6.show()
```
